## 2. divide the narrative into units of words.

```python
import re
from nltk.tokenize import sent_tokenize, word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer

stop_words = set(stopwords.words('english'))
stemmer    = PorterStemmer()

def preprocess(text):
    """Tokenize, lowercase, remove non-word tokens and stopwords, then stem."""
    words = [
        stemmer.stem(word.lower())
        for sentence in sent_tokenize(text)
        for word in word_tokenize(sentence)
        if re.match(r'^\w+$', word) and word.lower() not in stop_words
    ]
    return words

def preprocess_sentences(text):
    """tokenizes text into words. lowercase."""
    return [
        [word.lower() for word in word_tokenize(sentence)
         if re.match(r'^\w+$', word)]
        for sentence in sent_tokenize(text)
    ]

story_words_all_root  = [preprocess(story) for story in story_all]

recall_words_all_root = [
    [preprocess(recall) for recall in story_recalls]
    for story_recalls in recall_all
]

print(story_words_all_root[0][:20])
print(f"Narrative 1, participant 1 recall words: {recall_words_all_root[0][0][:20]}")
```

```python
# group them by cosine distance.

from sklearn.metrics.pairwise import cosine_similarity
import networkx as nx

threshold = np.cos(np.radians(60))  # cos(60°) = 0.5

story_word_groups = [[] for _ in range(12)]

for i in range(12):
    words = story_words_all_root[i]
    if len(words) == 0:
        continue

    # embed unique stemmed words
    unique_words = list(set(words))
    embeddings = model.encode(unique_words)

    # pairwise cosine similarity
    sim_matrix = cosine_similarity(embeddings)

    # build graph: connect words with similarity >= threshold
    G = nx.Graph()
    G.add_nodes_from(range(len(unique_words)))
    for a in range(len(unique_words)):
        for b in range(a+1, len(unique_words)):
            if sim_matrix[a, b] >= threshold:
                G.add_edge(a, b)

    # connected components = groups
    groups = [[unique_words[idx] for idx in component]
              for component in nx.connected_components(G)]
    story_word_groups[i] = groups

    print(f"Narrative {i+1}: {len(groups)} groups from {len(unique_words)} unique words")
```

```python
threshold = np.cos(np.radians(60))

freq_all = []
recall_rate_all = []

for i in range(12):
    story_words = story_words_all_root[i]
    if not story_words:
        continue

    unique_words = list(set(story_words))
    word_freq = {w: story_words.count(w) for w in unique_words}

    # embed unique story words
    story_emb = model.encode(unique_words)

    # recall rate for each unique word
    participant_rates = []
    for participant_words in recall_words_all_root[i]:
        if not participant_words:
            continue
        recall_emb = model.encode(participant_words)
        sim_matrix = cosine_similarity(story_emb, recall_emb)
        recalled = (np.max(sim_matrix, axis=1) >= threshold).astype(float)
        participant_rates.append(recalled)

    if not participant_rates:
        continue

    avg_recall = np.mean(participant_rates, axis=0)

    for j, word in enumerate(unique_words):
        freq_all.append(word_freq[word])
        recall_rate_all.append(avg_recall[j])
```

```python
## calculate per-participant recall-rate for SE estimate.

threshold = np.cos(np.radians(60))

freq_all = []
recall_rate_all = []
freq_participant_rate_all = []   # NEW: list of (frequency, participant's own mean recall at that frequency)

for i in range(12):
    story_words = story_words_all_root[i]
    if not story_words:
        continue

    unique_words = list(set(story_words))
    word_freq = {w: story_words.count(w) for w in unique_words}

    # embed unique story words
    story_emb = model.encode(unique_words)

    # recall matrix: rows = participants, columns = unique_words
    participant_rates = []
    for participant_words in recall_words_all_root[i]:
        if not participant_words:
            continue
        recall_emb = model.encode(participant_words)
        sim_matrix = cosine_similarity(story_emb, recall_emb)
        recalled = (np.max(sim_matrix, axis=1) >= threshold).astype(float)
        participant_rates.append(recalled)

    if not participant_rates:
        continue

    participant_rates = np.array(participant_rates)   # shape (n_participants, n_unique_words)
    avg_recall = participant_rates.mean(axis=0)        # per-word recall rate (unchanged from before)

    for j, word in enumerate(unique_words):
        freq_all.append(word_freq[word])
        recall_rate_all.append(avg_recall[j])

    # ── NEW: each participant's mean recall across this story's words, grouped by frequency ──
    word_freqs = np.array([word_freq[w] for w in unique_words])   # aligned with participant_rates columns
    for f in np.unique(word_freqs):
        col_idx = np.where(word_freqs == f)[0]
        participant_means_at_f = participant_rates[:, col_idx].mean(axis=1)  # one value per participant
        for pm in participant_means_at_f:
            freq_participant_rate_all.append((f, pm))
```

```python
# ── bin by frequency, mean/SE from PER-PARTICIPANT rates ────────────────────
freq_all = np.array(freq_all)
recall_rate_all = np.array(recall_rate_all)
freq_participant_rate_all = np.array(freq_participant_rate_all)  # (N, 2): col0=freq, col1=participant mean

freq_col = freq_participant_rate_all[:, 0]
rate_col = freq_participant_rate_all[:, 1]

max_freq = int(np.percentile(freq_all, 99.9999))  # ignore extreme outliers
bin_means, bin_sems, bin_counts, bin_n = [], [], [], []

for f in range(1, max_freq + 1):
    vals = rate_col[freq_col == f]
    n = len(vals)
    if n > 1:
        bin_means.append(np.mean(vals))
        bin_sems.append(np.std(vals, ddof=1) / np.sqrt(n))
        bin_counts.append(f)
        bin_n.append(n)
    elif n == 1:
        bin_means.append(vals[0])
        bin_sems.append(np.nan)   # SE undefined with a single participant
        bin_counts.append(f)
        bin_n.append(1)
```

```python
from scipy.stats import t as sp_t, linregress
from scipy.optimize import curve_fit
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.ticker import FuncFormatter

# font sizes: title / axis labels >= 30, everything else >= 25
fs_title, fs_label, fs_tick, fs_legend = 32, 30, 25, 25

weighted_fit = True          # weight the log fit by each bin's SE

x = np.array(bin_counts, dtype=float)
y = np.array(bin_means,  dtype=float)
se = np.array(bin_sems,  dtype=float)

# ── regression (kept for the r/p stats only, not plotted) ──────────────────
sl, ic, r_p, p_param, _ = linregress(x, y)

def add_stars(p):
    if p < 0.001: return '***'
    elif p < 0.01: return '**'
    elif p < 0.05: return '*'
    else: return 'ns'

def p_fmt(p):
    return "p < 0.001" if p < 0.001 else f"p = {p:.3f}"

# ── logarithmic fit: y = a + b * ln(x) ───────────────────────────────────────
def log_curve(xx, a, b):
    return a + b * np.log(xx)

ok = (x > 0) & ~np.isnan(y)
sigma = None
if weighted_fit:
    sigma = np.where(np.isnan(se) | (se <= 0), np.nanmax(se), se)[ok]

# start from the ordinary least-squares line in ln(x), then refine with the weights
b0, a0 = np.polyfit(np.log(x[ok]), y[ok], 1)
(a_fit, b_fit), cov = curve_fit(log_curve, x[ok], y[ok], p0=[a0, b0],
                                sigma=sigma, absolute_sigma=False, maxfev=10000)
a_err, b_err = np.sqrt(np.diag(cov))

# r and p of the log fit
n_fit  = int(ok.sum())
y_pred = log_curve(x[ok], a_fit, b_fit)
r2_log = 1 - np.sum((y[ok] - y_pred) ** 2) / np.sum((y[ok] - y[ok].mean()) ** 2)
r_log  = np.sign(b_fit) * np.sqrt(max(r2_log, 0))
t_log  = b_fit / b_err
p_log  = 2 * sp_t.sf(abs(t_log), df=n_fit - 2)

print(f"Log fit: y = {a_fit:.4f} + {b_fit:.4f} * ln(x)  "
      f"(a ± {a_err:.4f}, b ± {b_err:.4f}), r = {r_log:.3f}, p = {p_log:.3e}")
print(f"Linear regression: r = {r_p:.3f}, p = {p_param:.3e}")

# ── raw CI (±95% CI across participants, per frequency bin) ─────────────────
raw_lower, raw_upper = [], []
for m, s_e, n_p in zip(bin_means, bin_sems, bin_n):
    if n_p > 1:
        t_crit_i = sp_t.ppf(0.975, df=n_p - 1)
        raw_lower.append(m - t_crit_i * s_e)
        raw_upper.append(m + t_crit_i * s_e)
    else:
        raw_lower.append(np.nan)
        raw_upper.append(np.nan)
raw_lower = np.array(raw_lower)
raw_upper = np.array(raw_upper)

# ── plot ──
fig, ax = plt.subplots(figsize=(14, 10))

ax.fill_between(x, raw_lower, raw_upper, color='#1a5276', alpha=0.10, label="95% CI (data)")
ax.errorbar(x, y, yerr=se,
            fmt='-o', color='#1a5276', lw=3, markersize=10,
            ecolor='#1a5276', elinewidth=2, capsize=5,
            label="Mean ± SE (participants)")

# log curve (equation shown in %, matching the y-axis)
x_smooth = np.linspace(x[ok].min(), x[ok].max(), 300)
sign = '+' if b_fit >= 0 else '-'
ax.plot(x_smooth, log_curve(x_smooth, a_fit, b_fit), '--', color='#c0392b', lw=3.5,
        label=rf"Log fit: $y = {100 * a_fit:.1f} {sign} {100 * abs(b_fit):.1f}\,\ln(x)$"
              f"\nr = {r_log:.2f}, {p_fmt(p_log)} {add_stars(p_log)}")

# invisible handle just to carry the regression stats into the legend
ax.plot([], [], ' ', label=f"Regression  r = {r_p:.2f}, p = {p_param:.2e} {add_stars(p_param)}")

ax.set_xlabel("Word frequency in narratives", fontsize=fs_label)
ax.set_ylabel("% Participants recalled\nthe words", fontsize=fs_label)
ax.set_title("Word Recall Rates by Frequency in Narratives",
             fontsize=fs_title, fontweight='bold', pad=28)
ax.tick_params(labelsize=fs_tick)
ax.yaxis.set_major_formatter(FuncFormatter(lambda val, _: f"{val*100:.0f}"))
ax.spines[['top', 'right']].set_visible(False)

ax.set_xticks(bin_counts)
ax.legend(fontsize=fs_legend, loc='lower right')
plt.tight_layout()
plt.savefig("word_recall_by_frequency.png", dpi=150, bbox_inches='tight')
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.ticker import FuncFormatter
from scipy.stats import t as sp_t, linregress

angles     = [55, 60, 65]
colors     = ['#2471a3', '#e74c3c', '#27ae60']
alpha_fill = 0.12

results_reg = {}

for angle in angles:
    thr = np.cos(np.radians(angle))

    freq_all, recall_rate_all = [], []

    for i in range(12):
        story_words = story_words_all_root[i]
        if not story_words:
            continue

        unique_words = list(set(story_words))
        word_freq    = {w: story_words.count(w) for w in unique_words}
        story_emb    = model.encode(unique_words)

        participant_rates = []
        for participant_words in recall_words_all_root[i]:
            if not participant_words:
                continue
            recall_emb = model.encode(participant_words)
            sim        = cosine_similarity(story_emb, recall_emb)
            recalled   = (np.max(sim, axis=1) >= thr).astype(float)
            participant_rates.append(recalled)

        if not participant_rates:
            continue

        avg_recall = np.mean(participant_rates, axis=0)
        for j, word in enumerate(unique_words):
            freq_all.append(word_freq[word])
            recall_rate_all.append(avg_recall[j])

    freq_all        = np.array(freq_all)
    recall_rate_all = np.array(recall_rate_all)

    max_freq   = int(np.percentile(freq_all, 99.9999))
    bin_means, bin_stds, bin_counts = [], [], []
    for f in range(1, max_freq + 1):
        vals = recall_rate_all[freq_all == f]
        if len(vals) > 0:
            bin_means.append(np.mean(vals))
            bin_stds.append(np.std(vals))
            bin_counts.append(f)

    x = np.array(bin_counts, dtype=float)
    y = np.array(bin_means,  dtype=float)

    sl, ic, r_p, p_param, se_sl = linregress(x, y)
    n      = len(x)
    y_hat  = sl * x + ic
    s_e    = np.sqrt(np.sum((y - y_hat)**2) / (n - 2))
    x_mean = np.mean(x)
    Sxx    = np.sum((x - x_mean)**2)
    t_crit = sp_t.ppf(0.975, df=n - 2)

    x_fine   = np.linspace(x.min(), x.max(), 300)
    y_fine   = sl * x_fine + ic
    ci_width = t_crit * s_e * np.sqrt(1/n + (x_fine - x_mean)**2 / Sxx)

    results_reg[angle] = dict(
        x=x, y=y, sl=sl, ic=ic, r=r_p, p=p_param, n=n,
        x_fine=x_fine, y_fine=y_fine,
        ci_lower=y_fine - ci_width, ci_upper=y_fine + ci_width,
    )
    print(f"Angle {angle}°: n_bins={n}, slope={sl:.4f}, r={r_p:.3f}, p={p_param:.3e}")

```
